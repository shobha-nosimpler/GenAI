# Assistant role vs System role

In OpenAI's GPT models, roles such as "assistant" and "system" play distinct functions within the prompt framework, especially in structured conversations. Here's an overview of the differences between these roles:

## Assistant
  - Role: The "assistant" role represents the AI model that interacts directly with the user, providing responses, answering questions, and guiding the user based on the given inputs.

  - Purpose: This role is focused on delivering helpful, contextually relevant, and coherent responses to user queries.

  - Examples: The assistant role typically generates text that directly addresses the user's request, such as providing information, performing a task, or engaging in a conversation.

## System
  - Role: The "system" role is used to set the overall behavior, instructions, or context for the conversation. It defines the rules and guidelines that the assistant should follow while interacting with the user.

  - Purpose: This role ensures that the assistant maintains a certain tone, adheres to specific instructions, or operates within predefined boundaries throughout the interaction.

  - Examples: System instructions might include directives like "speak in a formal tone," "provide brief responses," or "focus on technical accuracy."

##Key Differences

  - Function: The system role sets the context and guidelines, while the assistant role executes the interaction based on those guidelines.

  - Focus: The system is about shaping the behavior of the assistant, whereas the assistant focuses on delivering user-facing responses.

  - By using these roles effectively, you can create structured and coherent interactions that adhere to specific requirements and maintain a consistent conversational style.

```python
max_conversations = 20
conversation_length = 0 # initialize

# initialize system message, user and assistant examples
message_history = [
        {"role": "system", "content": system_message},
        {"role": "system", "name":"example_user", "content": "Help me solve the equation 3x - 9 = 21."},
        {"role": "system", "name":"example_assistant", "content": "Sure! Try moving the 9 to the right hand side of the equation. What do you get?"},
        {"role": "system", "name":"example_user", "content": "3x = 12"},
        {"role": "system", "name":"example_assistant", "content": "Well, there seems to be a mistake. When you move 9 to the right hand side, you need to change its sign. Can you try again?"},
        {"role": "system", "name":"example_user", "content": "3x = 30"},
        {"role": "system", "name":"example_assistant", "content": "That looks good, great job! Now, try to divide both sides by 3. What do you get?"},
        {"role": "system", "name":"example_user", "content": "x = 10"}]
```

```python
# Taking user input for interactive conversation between Tutor and end user

while conversation_length < max_conversations:
  user_input = input()

  # exit if user enters exit
  if "exit" in user_input.lower():
    print("AI Tutor: Exiting the program!")
    break

  message_history.append({"role": "user", "content": user_input})

  chat_response = openai.chat.completions.create(
  model="gpt-3.5-turbo-16k",
  messages = message_history).choices[0].message.content

  print("\n", "AI Tutor:")
  print(chat_response)
  print("\n")

  message_history.append({"role": "assistant", "content": chat_response}) # add API response to message history
  conversation_length = conversation_length + 1 # another conversation done
```
